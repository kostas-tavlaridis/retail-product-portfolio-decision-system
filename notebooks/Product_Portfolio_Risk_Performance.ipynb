  <!-- TECHNICAL ABSTRACT -->
  <div style="font-size:22px;font-weight:700;color:#17263c;margin-bottom:14px;">Technical Abstract</div>
  <p style="font-size:15px;line-height:1.75;color:#2b3440;margin:0 0 14px 0;">This notebook documents the analytical system behind the <b>Product Portfolio Decision System</b>. Starting from SQL-cleaned transaction data, the analysis reconstructs a defensible <b>Order × Product</b> base and builds a <b>1,894-product analytical portfolio</b>. Each product is evaluated across four economic dimensions: <b>profitability, volatility, loss frequency, and economic scale</b>.</p>
  <p style="font-size:15px;line-height:1.75;color:#2b3440;margin:0 0 14px 0;">The resulting KPI geometry separates the <b>clusterable portfolio body</b> from structurally extreme observations, producing <b>four recurring body roles</b> and <b>three structural tail regimes</b>. These seven analytical roles define baseline management direction. Governance signals covering <b>deterioration, concentration, structural overlap, and assignment instability</b> then test whether that direction still holds. The evidence is synthesized into one <b>escalation status</b> and one <b>Final Intervention</b> for every product, while the same decisions are aggregated into four executive governance states.</p>
  <p style="font-size:15px;line-height:1.75;color:#2b3440;margin:0 0 22px 0;">A final decision layer compares actual product performance with <b>role-relative expectations</b>, applies explicit reliability guards, and isolates the cases that warrant additional management attention through five active <b>Priority Zones</b>. Robustness checks are attached to the analytical layer they test, distinguishing stable structural findings from threshold-sensitive management overlays.</p>

  <p style="
      font-size: 15px;
      line-height: 1.75;
      color: #2b3440;
      margin: 0 0 14px 0;
  ">
    The SQL-cleaned source is still a <b>transaction-line dataset</b>: each row
    represents one product entry recorded within an order. Because the same
    product can appear more than once within the same order, these rows cannot
    be treated as separate order-level outcomes.
  </p>

  <p style="
      font-size: 15px;
      line-height: 1.75;
      color: #2b3440;
      margin: 0 0 14px 0;
  ">
    We therefore first combine repeated lines belonging to the same
    <b>order and product</b>. Sales and profit are summed so that each row now
    represents one product's complete economic result within one order. This is
    the <b>Order × Product base</b>, and it provides the correct foundation for
    measuring how a product performs across different orders.
  </p>

  <p style="
      font-size: 15px;
      line-height: 1.75;
      color: #2b3440;
      margin: 0 0 18px 0;
  ">
    The Order × Product base is then aggregated to the <b>product level</b>.
    Each product is represented by its total sales, total profit, profit margin,
    average order profit, and the share of orders that resulted in a loss.
    The working product key is <b>Product ID × Product Name</b>, because Product ID
    alone is not fully unique in this dataset. This product-level diagnostic base
    becomes the starting point for all downstream portfolio analysis.
  </p>

          <td style="padding:10px 11px;border:1px solid #dce4ee;">
            <code>std(order profit) / (|mean order profit| + ε)</code>
          </td>
        </tr>
        <tr>
          <td style="padding:10px 11px;border:1px solid #dce4ee;">
            <b>Loss Rate</b>
          </td>
          <td style="padding:10px 11px;border:1px solid #dce4ee;">
            How frequently the product produces a loss when it appears in an order.
          </td>
          <td style="padding:10px 11px;border:1px solid #dce4ee;">
            <code>loss-making order-product observations / all order-product observations</code>
          </td>
        </tr>
        <tr style="background:#f8fafc;">
          <td style="padding:10px 11px;border:1px solid #dce4ee;">
            <b>Profit Scale</b>
          </td>
          <td style="padding:10px 11px;border:1px solid #dce4ee;">
            The product's economic contribution relative to the largest absolute product-profit outcome in the portfolio.
          </td>
          <td style="padding:10px 11px;border:1px solid #dce4ee;">
            <code>total product profit / max(|product profit|)</code>
          </td>
        </tr>
      </tbody>
    </table>

  </div>

</div>

```python
# =============================================================================
# 3.1 CORE PRODUCT METRICS
# =============================================================================

# Core metrics, CV stabilization and the KPI profile share one validated run.
chapter_03 = run_chapter_03(
    analysis_config, chapter_01.order_prod, chapter_01.totals,
    show_audit=False, show_preview=False,
)
vol = chapter_03.vol
```

  ">
      3.2 Volatility Treatment
  </div>

  <p style="
      font-size:15px;
      line-height:1.75;
      color:#2b3440;
      margin:0 0 14px 0;
  ">
    Starting from the product-level order statistics calculated above, the
    pipeline measures volatility by comparing each product's
    <b>standard deviation of order profit</b> with its
    <b>average order profit</b>. This produces the coefficient of variation (CV).
  </p>

  <div style="
      background:#f7f9fc;
      border:1px solid #dce4ee;
      border-radius:7px;
      padding:12px 16px;
      margin:14px 0 18px 0;
      text-align:center;
      font-size:15px;
      color:#263444;
  ">
      <b>CV = standard deviation of order profit / (|mean order profit| + ε)</b>
  </div>

  <p style="
      font-size:15px;
      line-height:1.75;
      color:#2b3440;
      margin:0 0 14px 0;
  ">
    The calculation is performed in four practical steps:
  </p>

  <div style="
      background:#f7f9fc;
      border:1px solid #dce4ee;
      border-radius:8px;
      padding:15px 18px;
      margin:16px 0 18px 0;
      font-size:14px;
      line-height:1.8;
      color:#334155;
  ">
    <b>1.</b> Calculate the number of observed orders, mean order profit, and
    standard deviation for every product.<br>
    <b>2.</b> Add a small denominator adjustment,
    <b>ε = 0.2533</b>, equal to the 1st percentile of absolute mean order
    profit. This prevents near-zero average profit from producing unstable
    ratios.<br>
    <b>3.</b> Preserve the resulting uncapped value in
    <code>cv_order_profit_raw</code> for interpretation and audit.<br>
    <b>4.</b> Create a separate
    <code>cv_order_profit_model</code> field for the distance-based geometry,
    applying explicit rules to extreme and non-estimable cases.
  </div>

  <p style="
      font-size:15px;
      line-height:1.75;
      color:#2b3440;
      margin:0 0 16px 0;
  ">
    The word <b>raw</b> therefore means <i>uncapped diagnostic CV</i>: the
    denominator adjustment has already been applied, but the observed ratio has
    not been restricted. The modelling field is the controlled coordinate used
    later in body/tail separation and K-means.
  </p>

        <tr>
          <td style="padding:9px 10px;">
            <b>Economic Scale</b>
          </td>
          <td style="padding:9px 10px;">
            The central P01–P99 range is approximately
            <b>−0.0317 to 0.0832</b>, while the maximum reaches
            <b>1.0000</b>.
          </td>
          <td style="padding:9px 10px;">
            A small number of products have economic impact far beyond the
            central portfolio scale.
          </td>
        </tr>
      </tbody>
    </table>
    <div style="
        font-size:14px;
        line-height:1.7;
        color:#4a4438;
        margin-top:13px;
    ">
      The diagnostic therefore supports testing a body/tail split on
      profitability, volatility, and economic scale. It does not by itself claim
      that 1%–99% is an optimal cutoff.
    </div>

  </div>

</div>

<div style="
    font-family:Arial, sans-serif;
    max-width:1150px;
    margin:0 auto;
">

  <div style="
      font-size:19px;
      font-weight:700;
      color:#244e7a;
      margin:27px 0 9px 0;
  ">
      4.2 Empirical Body / Tail Separation
  </div>

  <p style="
      font-size:15px;
      line-height:1.75;
      color:#2b3440;
      margin:0 0 14px 0;
  ">
    The body is defined directly from the observed portfolio distribution rather
    than through a Gaussian rule such as ±3 standard deviations. Normal-theory
    filtering would be inappropriate for a KPI space combining skewed
    profitability, ratio-based volatility, bounded loss frequency, and
    heavy-tailed economic impact.
  </p>

  <p style="
      font-size:15px;
      line-height:1.75;
      color:#2b3440;
      margin:0 0 18px 0;
  ">
    The selected operational rule retains products between the
    <b>1st and 99th percentiles</b> of Mean Order Profit, model CV, and Profit
    Scale. A product must remain inside all three ranges to enter the body.
    Falling outside any one range routes it to the structural-tail candidate set.
  </p>


  <div style="
      background:#f7f9fc;
      border:1px solid #dce4ee;
      border-radius:8px;
      padding:15px 17px;
      margin:18px 0;
  ">
    <table style="
        width:100%;
        border-collapse:collapse;
        font-size:13.5px;
        color:#263444;
    ">
      <thead>
        <tr style="background:#17263c;color:white;">
          <th style="padding:9px 11px;text-align:left;border:1px solid #17263c;width:23%;">
            Layer
          </th>
          <th style="padding:9px 11px;text-align:left;border:1px solid #17263c;width:40%;">
            Definition
          </th>
          <th style="padding:9px 11px;text-align:left;border:1px solid #17263c;">
            Downstream Treatment
          </th>
        </tr>
      </thead>
      <tbody>
        <tr>
          <td style="padding:10px 11px;border:1px solid #dce4ee;">
            <b>Portfolio Body</b>
          </td>
          <td style="padding:10px 11px;border:1px solid #dce4ee;">
            Inside all three selected empirical boundaries.
          </td>
          <td style="padding:10px 11px;border:1px solid #dce4ee;">
            Passed to K-means to identify recurring structural roles.
          </td>
        </tr>
        <tr style="background:#f8fafc;">
          <td style="padding:10px 11px;border:1px solid #dce4ee;">
            <b>Structural Tail</b>
          </td>
          <td style="padding:10px 11px;border:1px solid #dce4ee;">
            Outside at least one selected empirical boundary.
          </td>
          <td style="padding:10px 11px;border:1px solid #dce4ee;">
            Preserved and later classified through explicit tail typology.
          </td>
        </tr>
      </tbody>
    </table>

  </div>


  <div style="
      background:#fffaf0;
      border-left:3px solid #b58a37;
      padding:13px 16px;
      margin:18px 0;
  ">
    <div style="
        font-size:13px;
        font-weight:700;
        color:#725821;
        margin-bottom:5px;
    ">
      Why Loss Rate is different
    </div>
    <div style="
        font-size:14px;
        line-height:1.7;
        color:#4a4438;
    ">
      Loss Rate remains inside the four-dimensional K-means space, but it is not
      used as an exclusion rule. Its P01–P99 range already spans its natural
      0–1 limits. Keeping it inside the body allows recurring downside behaviour
      to help differentiate clusters without automatically treating every
      0% or 100% loss-rate product as a structural extreme.
    </div>
  </div>

</div>

```python
# =============================================================================
# 4.2 EMPIRICAL BODY / TAIL SEPARATION
# =============================================================================

body_boundary_specs = chapter_04.body_boundary_specs
body_thresholds_df = chapter_04.body_thresholds_df
body_mask = chapter_04.body_mask
km_df_clust = chapter_04.km_df_clust
km_df_tail_candidates = chapter_04.km_df_tail_candidates

total_modelling_products = len(chapter_04.km_df_raw)
body_products = len(chapter_04.km_df_clust)
tail_products = len(chapter_04.km_df_tail_candidates)
body_retention_rate = body_products / total_modelling_products
structural_tail_share = tail_products / total_modelling_products

body_thresholds_view = (
    body_thresholds_df[
        [
            "feature",
            "lower_quantile",
            "upper_quantile",
            "lower_value",
            "upper_value",
        ]
    ]
    .copy()
    .rename(columns={
        "feature": "KPI",
        "lower_quantile": "Lower Quantile",
        "upper_quantile": "Upper Quantile",
        "lower_value": "Lower Value",
        "upper_value": "Upper Value",
    })
)

body_thresholds_view["KPI"] = (
    body_thresholds_view["KPI"]
    .map({
        "mean_order_profit": "Mean Order Profit",
        "cv_order_profit": "Model CV",
        "profit_scale": "Profit Scale",
    })
)

print("\n=== SELECTED EMPIRICAL BODY BOUNDARIES ===")

display(
    body_thresholds_view.round({
        "Lower Quantile": 3,
        "Upper Quantile": 3,
        "Lower Value": 4,
        "Upper Value": 4,
    })
)

print("\n=== EMPIRICAL BODY / TAIL SEPARATION SUMMARY ===")

print(f"Products before separation:  {total_modelling_products:,}")

print(f"Products retained in body:   {body_products:,}")

print(f"Structural-tail candidates:  {tail_products:,}")

print(f"Body retention rate:         {body_retention_rate:.1%}")

print(f"Structural-tail share:       {structural_tail_share:.1%}")
```

      Separation result

    <div style="

      The selected rule retains <b>1,822 products (96.2%)</b> inside the

  <div style="
      font-size:19px;
      font-weight:700;
      color:#244e7a;
      margin:27px 0 9px 0;
  ">
      4.3 Boundary Sensitivity
  </div>

  <p style="
      font-size:15px;
      line-height:1.75;
      color:#2b3440;
      margin:0 0 14px 0;
  ">
    The 1%–99% rule is a documented analytical choice, not a mathematically
    optimized cutoff. It is therefore compared with both a
    <b>no-separation control</b> and nearby alternatives ranging from a broad
    <b>0.5%–99.5%</b> body to a tighter <b>5%–95%</b> body.
  </p>

  <div style="
      background:#fffaf0;
      border-left:3px solid #b58a37;
      padding:13px 16px;
      margin:16px 0 18px 0;
  ">
    <div style="
        font-size:13px;
        font-weight:700;
        color:#725821;
        margin-bottom:5px;
    ">
      What the 0%–100% test means
    </div>
    <div style="
        font-size:14px;
        line-height:1.7;
        color:#4a4438;
    ">
      The <b>0%–100%</b> case uses the observed minimum and maximum of every
      boundary feature, so all <b>1,894 products</b> remain inside K-means and
      no structural tail is created. It is not a candidate final rule. It is the
      control case that shows what the same K=4 model does when body/tail
      separation is switched off completely.
    </div>
  </div>

  <p style="
      font-size:15px;
      line-height:1.75;
      color:#2b3440;
      margin:0 0 14px 0;
  ">
    Every case is evaluated under the same standardized
    <b>K=4 reference structure</b>. Holding K, the random seed, and model
    initialization constant ensures that changes are attributable to the
    boundary definition rather than to a different clustering setup.
  </p>

  <p style="
      font-size:15px;
      line-height:1.75;
      color:#2b3440;
      margin:0 0 14px 0;
  ">
    The comparison uses <b>silhouette</b> to measure group separation and the
    <b>smallest cluster size</b> as an important guard. Silhouette alone can look
    strong when a few distant observations form a tiny isolated cluster. The
    additional size check reveals whether K-means is describing recurring
    portfolio structure or spending an entire cluster on only a handful of
    extremes.
  </p>

  <p style="
      font-size:15px;
      line-height:1.75;
      color:#2b3440;
      margin:0 0 18px 0;
  ">
    <b>Inertia is not used</b> for this boundary comparison. It totals product
    distances from their assigned centres and therefore changes partly because
    tighter rules retain fewer products. The decision is read directly through
    coverage, tail size, silhouette, and cluster utilization; no weighted score
    or mechanical optimum is imposed.
  </p>

</div>

```python
# =============================================================================
# 4.3 BODY-BOUNDARY SENSITIVITY
# =============================================================================

body_boundary_sensitivity = chapter_04.body_boundary_sensitivity

boundary_sensitivity_view = body_boundary_sensitivity.copy()

boundary_sensitivity_view["Boundary"] = (
    boundary_sensitivity_view["boundary_case"]
    .replace({
        "No separation": "No separation (0%–100%)",
    })
)

boundary_sensitivity_view["Body Products"] = (
    boundary_sensitivity_view["body_products"]
    .map(lambda value: f"{value:,}")
)

boundary_sensitivity_view["Tail Products"] = (
    boundary_sensitivity_view["tail_products"]
    .map(lambda value: f"{value:,}")
)

boundary_sensitivity_view["Body Retention"] = (
    boundary_sensitivity_view["body_retention_rate"]
    .map(lambda value: f"{value:.1%}")
)

boundary_sensitivity_view["Smallest K=4 Cluster"] = (
    boundary_sensitivity_view["smallest_cluster_products_k4"]
    .astype(int)
    .map(lambda value: f"{value:,}")
    + " ("
    + boundary_sensitivity_view["smallest_cluster_share_k4"]
    .map(lambda value: f"{value:.1%}")
    + ")"
)

boundary_sensitivity_view["Silhouette (K=4)"] = (
    boundary_sensitivity_view["silhouette_score_k4"]
    .map(lambda value: f"{value:.3f}")
)

boundary_sensitivity_view["Change vs No Split"] = (
    boundary_sensitivity_view["silhouette_change_vs_no_separation"]
    .map(
        lambda value: "0.000"
        if abs(value) < 0.0005
        else f"{value:+.3f}"
    )
)

boundary_sensitivity_view["Selected"] = (
    boundary_sensitivity_view["selected_boundary"]
    .map({True: "✓", False: ""})
)

print("\n=== BODY-BOUNDARY SENSITIVITY CHECK ===")

display(
    boundary_sensitivity_view[
        [
            "Boundary",
            "Body Products",
            "Tail Products",
            "Body Retention",
            "Smallest K=4 Cluster",
            "Silhouette (K=4)",
            "Change vs No Split",
            "Selected",
        ]
    ]
)
```

    <div style="font-size:14px;line-height:1.7;color:#4a4438;">The no-separation case produces the highest overall silhouette, but one cluster contains only one product.           <td style="padding:10px 11px;border:1px solid #dce4ee;"><b>Profiling</b></td>
          <td style="padding:10px 11px;border:1px solid #dce4ee;">What economic behaviour does each validated cluster represent?</td>
          <td style="padding:10px 11px;border:1px solid #dce4ee;">Raw-KPI cluster profiles and role names</td>
        </tr>
      </tbody>
    </table>
  </div>
</div>


<div style="font-family:Arial, sans-serif;max-width:1150px;margin:0 auto;">
  <div style="font-size:19px;font-weight:700;color:#244e7a;margin:27px 0 9px 0;">
    5.1 Comparable Distance Space
  </div>

  <p style="font-size:15px;line-height:1.75;color:#2b3440;margin:0 0 14px 0;">
    K-means groups products by distance. In the original KPI units, a feature
    with a wider numerical range can dominate that distance simply because its
    numbers are larger. That would give the algorithm an unintended preference,
    not an economically justified one.
  </p>

  <p style="font-size:15px;line-height:1.75;color:#2b3440;margin:0 0 18px 0;">
    <code>StandardScaler</code> therefore recentres each KPI around zero and
    expresses it in standard-deviation units. This does <b>not</b> change the
    product ranking or the meaning of the KPI. It only places Mean Order Profit,
    CV, Loss Rate, and Profit Scale on comparable terms for the distance calculation.
  </p>

  <div style="background:#fffaf0;border-left:3px solid #b58a37;padding:13px 16px;margin:18px 0;">
    <div style="font-size:13px;font-weight:700;color:#725821;margin-bottom:5px;">Interpretation boundary</div>
    <div style="font-size:14px;line-height:1.7;color:#4a4438;">
      Scaling is used only to fit K-means. Cluster profiles and role names are
      calculated afterward from the <b>original KPI units</b>, so the results
      remain economically readable.
    </div>
  </div>
</div>


```python
# =============================================================================
# 5.1 SCALE THE CLUSTERABLE BODY
# =============================================================================

chapter_05 = run_chapter_05(
    analysis_config, chapter_03.df_seg, chapter_04, show_outputs=False,
)
scaler = chapter_05.scaler
X_scaled = chapter_05.x_scaled
scaling_audit = chapter_05.scaling_audit

print("\n=== SCALING AUDIT — CLUSTERABLE BODY ===")

display(
    scaling_audit.round({
        "Raw Mean": 4,
        "Raw Std": 4,
        "Scaled Mean": 4,
        "Scaled Std": 4,
    })
)
```

<div style="font-family:Arial, sans-serif;max-width:1150px;margin:0 auto;">
  <div style="font-size:19px;font-weight:700;color:#244e7a;margin:27px 0 9px 0;">
    5.2 Selecting the Number of Structures
  </div>

  <p style="font-size:15px;line-height:1.75;color:#2b3440;margin:0 0 14px 0;">
    Candidate solutions from <b>K=2 to K=8</b> are compared under the same
    scaled body. Two diagnostics answer different questions. <b>Inertia</b>
    asks how tightly products sit around their group centres; because it always
    falls as more groups are added, we look for the point where the extra gain
    begins to flatten. <b>Silhouette</b> asks whether products are closer to their
    own group than to neighbouring groups; higher values indicate clearer separation.
  </p>

  <p style="font-size:15px;line-height:1.75;color:#2b3440;margin:0 0 18px 0;">
    Here inertia is valid because every candidate uses the <b>same 1,822 products</b>.
    The final choice is not made by one score alone: the diagnostics narrow the
    options, while the raw-KPI profiles determine whether the added groups describe
    distinct and useful economic regimes.
  </p>
</div>


```python
# =============================================================================
# 5.2 K SELECTION — ELBOW AND SILHOUETTE
# =============================================================================

# Presentation only: model selection was computed in section 5.1.
K_OPT = analysis_config.k_opt
kmeans_model_selection = chapter_05.kmeans_model_selection

print("\n=== K-MEANS MODEL-SELECTION EVIDENCE ===")

display(
    kmeans_model_selection.round({
        "inertia": 1,
        "silhouette_score": 3,
    })
)

fig, axes = plt.subplots(1, 2, figsize=(12, 4.2))

axes[0].plot(
    kmeans_model_selection["K"],
    kmeans_model_selection["inertia"],
    marker="o",
    color="#244e7a",
)
axes[0].axvline(K_OPT, color="#4f9c70", linestyle="--", linewidth=1.4)
axes[0].set_title("Elbow — Within-Group Distance")
axes[0].set_xlabel("Number of clusters (K)")
axes[0].set_ylabel("Inertia")
axes[0].grid(alpha=0.25)

silhouette_view = kmeans_model_selection.dropna(
    subset=["silhouette_score"]
)

axes[1].plot(
    silhouette_view["K"],
    silhouette_view["silhouette_score"],
    marker="o",
    color="#8a5a9f",
)
axes[1].axvline(K_OPT, color="#4f9c70", linestyle="--", linewidth=1.4)
axes[1].set_title("Silhouette — Group Separation")
axes[1].set_xlabel("Number of clusters (K)")
axes[1].set_ylabel("Silhouette score")
axes[1].grid(alpha=0.25)

plt.tight_layout()
plt.show()
plt.close(fig)
```

<div style="font-family:Arial, sans-serif;max-width:1150px;margin:18px auto 0 auto;">
  <div style="background:#f1faf5;border-left:3px solid #4f9c70;padding:14px 17px;">
    <div style="font-size:13px;font-weight:700;color:#31714d;margin-bottom:5px;">K decision</div>
    <div style="font-size:14px;line-height:1.75;color:#34483c;">
      <b>K=4</b> is retained as the operating segmentation. The elbow begins to
      flatten after four groups, while the selected solution retains a
      <b>0.557 silhouette</b> and produces four distinct economic profiles.
      K=3 compresses roles that behave differently, whereas K=5 adds
      fragmentation without a correspondingly useful new portfolio regime.
      This is therefore a documented analytical choice, not a claim that four
      is the only mathematically possible partition.
    </div>
  </div>
</div>


<div style="font-family:Arial, sans-serif;max-width:1150px;margin:0 auto;">
  <div style="font-size:19px;font-weight:700;color:#244e7a;margin:27px 0 9px 0;">
    5.3 Final Body Assignment
  </div>

  <p style="font-size:15px;line-height:1.75;color:#2b3440;margin:0 0 14px 0;">
    The final model is fitted with <b>K=4</b>, a fixed random seed, and
    <code>n_init=20</code>. Multiple initializations reduce the chance that the
    result depends on one poor starting position.
  </p>

  <p style="font-size:15px;line-height:1.75;color:#2b3440;margin:0 0 14px 0;">
    Cluster labels are attached only to the central body. Tail products are
    marked <code>Unclustered</code>; this is a routing status, not missing data
    and not evidence that those products were discarded.
  </p>

  <p style="font-size:15px;line-height:1.75;color:#2b3440;margin:0 0 18px 0;">
    The numerical assignments are now available. Before interpreting and naming
    the clusters, the next section tests whether this structure survives
    reasonable changes in sample composition and CV support.
  </p>
</div>


```python
# =============================================================================
# 5.3 FIT THE FINAL K-MEANS MODEL AND RESTORE PORTFOLIO CONTEXT
# =============================================================================

kmeans = chapter_05.kmeans
labels = chapter_05.labels
km_df_clust = chapter_05.km_df_clust
df_seg_clusters = chapter_05.df_seg_clusters
assignment_summary = chapter_05.assignment_summary

print("\n=== FINAL K-MEANS ASSIGNMENT SUMMARY ===")

display(
    assignment_summary.assign(
        portfolio_share=assignment_summary["portfolio_share"].map(
            lambda value: f"{value:.1%}"
        )
    )
)
```

<div style="font-family:Arial, sans-serif;max-width:1150px;margin:0 auto;">
  <div style="font-size:19px;font-weight:700;color:#244e7a;margin:27px 0 9px 0;">
    5.4 Stability and Support Sensitivity
  </div>

  <p style="font-size:15px;line-height:1.75;color:#2b3440;margin:0 0 14px 0;">
    Before assigning economic meaning to the numerical clusters, we test
    whether the underlying grouping is stable. A useful segmentation should not
    disappear when the input changes slightly. Two separate checks are used.
    First, the model is refitted on <b>100 random 80% subsamples</b>. Second, it
    is refitted after excluding body products whose CV is not estimable because
    they have fewer than two observations.
  </p>

  <p style="font-size:15px;line-height:1.75;color:#2b3440;margin:0 0 18px 0;">
    Agreement is measured with the <b>Adjusted Rand Index (ARI)</b>. ARI compares
    whether products remain grouped together and ignores arbitrary changes in
    cluster numbers. A value near 1 means that the same basic structure reappears.
    Silhouette is tracked alongside it to confirm that separation quality also
    remains consistent.
  </p>
</div>


```python
# =============================================================================
# 5.4 CLUSTER STABILITY AND CV-SUPPORT SENSITIVITY
# =============================================================================

cluster_stability_iterations = chapter_05.cluster_stability_iterations
cluster_stability_summary = chapter_05.cluster_stability_summary
cv_support_summary = chapter_05.cv_support_summary

print("\n=== RESAMPLING STABILITY ===")

display(cluster_stability_summary.round(3))

print("\n=== NON-ESTIMABLE-CV SUPPORT SENSITIVITY ===")

display(cv_support_summary.round(3))
```

<div style="font-family:Arial, sans-serif;max-width:1150px;margin:18px auto 0 auto;">
  <div style="background:#f1faf5;border-left:3px solid #4f9c70;padding:14px 17px;">
    <div style="font-size:13px;font-weight:700;color:#31714d;margin-bottom:5px;">Validation result</div>
    <div style="font-size:14px;line-height:1.75;color:#34483c;">
      The four-cluster structure is robust on both tests. Across the 100
      resamples, median ARI is <b>0.976</b> with a 5th–95th percentile range
      of <b>0.944–0.995</b>. Excluding the 80 non-estimable-CV products produces
      an ARI of <b>0.992</b>, while silhouette remains effectively unchanged
      (<b>0.557 → 0.558</b>). The numerical structure is therefore not a fragile
      result of one sample or of the low-support CV modelling coordinate and can
      now be interpreted economically.
    </div>
  </div>
</div>


<div style="font-family:Arial, sans-serif;max-width:1150px;margin:0 auto;">

    5.5 Cluster Profiles and Economic Roles

  </p>

```python
# =============================================================================
# 5.5 PROFILE AND NAME THE FOUR BODY CLUSTERS
# =============================================================================

cluster_profile = chapter_05.cluster_profile

cluster_profile_view = cluster_profile.copy()

cluster_profile_view["Portfolio Footprint"] = (
    cluster_profile_view["pct_products"]
    .map(lambda value: f"{value:.1f}%")
)

cluster_profile_view["Net Profit Share"] = (
    cluster_profile_view["pct_net_profit"]
    .map(lambda value: f"{value:.1f}%")
)

cluster_profile_view["Total Profit"] = (
    cluster_profile_view["total_profit"]
    .map(lambda value: f"{value:,.2f}")
)

cluster_profile_view["Mean Order Profit"] = (
    cluster_profile_view["mean_order_profit"]
    .map(lambda value: f"{value:,.2f}")
)

cluster_profile_view["Median CV"] = (
    cluster_profile_view["median_cv"]
    .map(lambda value: f"{value:.2f}")
)

cluster_profile_view["Mean Loss Rate"] = (
    cluster_profile_view["mean_loss_rate"]
    .map(lambda value: f"{value:.1%}")
)

print("\n=== K-MEANS BODY ROLE PROFILE ===")

display(
    cluster_profile_view[
        [
            "body_role",
            "k_cluster",
            "n_products",
            "Portfolio Footprint",
            "Total Profit",
            "Net Profit Share",
            "Mean Order Profit",
            "Median CV",
            "Mean Loss Rate",
        ]
    ]
    .rename(columns={
        "body_role": "Economic Role",
        "k_cluster": "Cluster",
        "n_products": "Products",
    })
)
```

<div style="font-family:Arial, sans-serif;max-width:1150px;margin:0 auto;">
  <div style="font-size:19px;font-weight:700;color:#244e7a;margin:27px 0 9px 0;">6.1 Body-Anchored Tail References</div>

  <p style="font-size:15px;line-height:1.75;color:#2b3440;margin:0 0 14px 0;">
    Tail severity is measured against the <b>clustered body</b>, not against the
    full portfolio. If extreme observations were allowed to set their own
    reference points, the thresholds would move outward and make exceptional
    behaviour appear more ordinary than it is.
  </p>

  <p style="font-size:15px;line-height:1.75;color:#2b3440;margin:0 0 18px 0;">
    Volatility and positive impact therefore use the body's 95th percentiles.
    Material loss uses a small negative Profit Scale boundary of <b>−0.001</b>,
    preventing negligible near-zero effects from being labelled as structural
    downside. The code also verifies that body and tail form a complete,
    non-overlapping partition before any rule is applied.
  </p>
</div>

```python
# =============================================================================
# 6.1 CONSTRUCT BODY-ANCHORED TAIL REFERENCES
# =============================================================================

chapter_06 = run_chapter_06(
    analysis_config, chapter_04, chapter_05, show_outputs=False,
)
body = chapter_06.body
tail = chapter_06.tail
tail_reference_table = chapter_06.tail_reference_table
tail_materiality = chapter_06.tail_materiality
cv_thr = chapter_06.cv_threshold
impact_thr = chapter_06.impact_threshold
loss_scale_tol = chapter_06.loss_scale_tolerance

print("\n=== BODY-ANCHORED TAIL REFERENCES ===")

display(tail_reference_table.round({"Observed Threshold": 4}))

print("\n=== BODY / TAIL MATERIALITY ===")

display(
    tail_materiality.assign(
        **{
            "Portfolio Share": tail_materiality["Portfolio Share"].map(
                lambda value: f"{value:.1%}"
            ),
            "Net Profit Share": tail_materiality["Net Profit Share"].map(
                lambda value: f"{value:.1%}"
            ),
            "Total Profit": tail_materiality["Total Profit"].map(
                lambda value: f"{value:,.2f}"
            ),
        }
    )
)
```

<div style="font-family:Arial, sans-serif;max-width:1150px;margin:0 auto;">
  <div style="font-size:19px;font-weight:700;color:#244e7a;margin:27px 0 9px 0;">6.2 Mechanism Flags and Primary Assignment</div>

  <p style="font-size:15px;line-height:1.75;color:#2b3440;margin:0 0 14px 0;">
    The three raw flags are intentionally allowed to overlap. A loss-making
    product can also be highly volatile, and a high-impact product can also be
    unstable. Preserving that overlap keeps the complete economic evidence.
  </p>

  <p style="font-size:15px;line-height:1.75;color:#2b3440;margin:0 0 18px 0;">
    Reporting still needs one primary class, so overlapping cases follow the
    fixed hierarchy <b>Loss → Volatility → Impact</b>. Value destruction comes
    first, instability comes next, and positive scale is used when neither of
    the more urgent mechanisms dominates. The raw flags remain unchanged.
  </p>

  <div style="background:#fffaf0;border-left:3px solid #b58a37;padding:13px 16px;margin:18px 0;">
    <div style="font-size:13px;font-weight:700;color:#725821;margin-bottom:5px;">Closed assignment without a blind fallback</div>
    <div style="font-size:14px;line-height:1.7;color:#4a4438;">
      The raw mechanisms directly cover <b>70 of 72 tail products (97.2%)</b>.
      The remaining two entered the tail only because Mean Order Profit exceeds
      the selected upper body boundary. They are assigned to
      <b>Extreme Impact Tail</b> through that observed positive direction, and the
      exact basis is recorded. Any case that fits neither a raw mechanism nor a
      documented boundary direction stops the pipeline instead of receiving a
      generic label.
    </div>
  </div>

  <div style="background:#f7f9fc;border:1px solid #dce4ee;border-radius:7px;padding:13px 16px;margin:18px 0;">
    <div style="font-size:13px;font-weight:700;color:#52657a;margin-bottom:5px;">Identity safeguard</div>
    <div style="font-size:14px;line-height:1.7;color:#3f4d5d;">
      Product ID alone is not unique in the source. Tail assignments are therefore
      written back through the preserved product-row index created at the
      <b>Product ID × Product Name</b> grain, preventing duplicate IDs from
      multiplying or losing records.
    </div>
  </div>
</div>

```python
# =============================================================================
# 6.2 APPLY RAW TAIL MECHANISMS AND THE PRIMARY-CLASS HIERARCHY
# =============================================================================

tail_entry_audit = chapter_06.tail_entry_audit
mechanism_coverage_summary = chapter_06.mechanism_coverage_summary
fallback_tail_cases = chapter_06.fallback_tail_cases
assignment_basis_summary = chapter_06.assignment_basis_summary
fallback_assignment_audit = chapter_06.fallback_assignment_audit
df_seg_clusters = chapter_06.df_seg_clusters

print("\n=== RAW-MECHANISM COVERAGE ===")

display(mechanism_coverage_summary)

print("\n=== TAIL PRODUCTS WITHOUT A RAW MECHANISM ===")

display(fallback_tail_cases)

print("\n=== PRIMARY TAIL ASSIGNMENT BASIS ===")

display(assignment_basis_summary)

print("\n=== BOUNDARY-DIRECTION ASSIGNMENT AUDIT ===")

display(fallback_assignment_audit)

print("\n=== TAIL ASSIGNMENT COMPLETED ===")

print(f"Body products:                  {len(body):,}")

print(f"Structural-tail products:      {len(tail):,}")

print(
    "Boundary-direction products:  "
    f"{int(tail['fallback_assignment_flag'].sum()):,}"
)

print(
    "Products in unified taxonomy: "
    f"{len(df_seg_clusters):,}"
)
```

<div style="font-family:Arial, sans-serif;max-width:1150px;margin:0 auto;">
  <div style="font-size:19px;font-weight:700;color:#244e7a;margin:27px 0 9px 0;">6.3 Tail Profile and Unified Taxonomy</div>

  <p style="font-size:15px;line-height:1.75;color:#2b3440;margin:0 0 14px 0;">
    Primary types make the tail readable, while the overlap audit shows how often
    a product carries more than one underlying mechanism. The final
    <code>analytical_role</code> combines the four body roles with the three tail
    types into one closed seven-role portfolio taxonomy.
  </p>

  <p style="font-size:15px;line-height:1.75;color:#2b3440;margin:0 0 18px 0;">
    The consolidation does not claim that body and tail roles were produced in
    the same way. Body roles come from recurring K-means geometry; tail roles
    come from explicit exception rules. Their source fields remain separate, so
    governance can distinguish a recurring operating regime from a structural
    exception at any point.
  </p>

  <div style="background:#f7f9fc;border:1px solid #dce4ee;border-radius:7px;padding:13px 16px;margin:18px 0;">
    <div style="font-size:13px;font-weight:700;color:#52657a;margin-bottom:5px;">Lean audit design</div>
    <div style="font-size:14px;line-height:1.7;color:#3f4d5d;">
      Detailed tables remain visible in the notebook, but only two reusable CSVs
      are written: one product-level tail audit containing thresholds, flags, and
      assignment provenance; and one complete seven-role portfolio profile.
    </div>
  </div>
</div>

```python
# =============================================================================
# 6.3 BUILD TAIL AUDITS AND THE SEVEN-ROLE ANALYTICAL TAXONOMY
# =============================================================================

tail_summary = chapter_06.tail_summary
overlap_audit = chapter_06.overlap_audit
tail_export = chapter_06.tail_export
analytical_role_profile = chapter_06.analytical_role_profile

tail_summary_view = tail_summary.copy()

tail_summary_view["Portfolio Footprint"] = (
    tail_summary_view["pct_products"]
    .map(lambda value: f"{value:.1f}%")
)

tail_summary_view["Total Profit"] = (
    tail_summary_view["total_profit"]
    .map(lambda value: f"{value:,.2f}")
)

tail_summary_view["Net Profit Share"] = (
    tail_summary_view["pct_net_profit"]
    .map(lambda value: f"{value:.1f}%")
)

tail_summary_view["Mean Order Profit"] = (
    tail_summary_view["mean_order_profit"]
    .map(lambda value: f"{value:,.2f}")
)

tail_summary_view["Median CV"] = (
    tail_summary_view["median_cv"]
    .map(lambda value: f"{value:.2f}")
)

tail_summary_view["Median Loss Rate"] = (
    tail_summary_view["median_loss_rate"]
    .map(lambda value: f"{value:.1%}")
)

print("\n=== STRUCTURAL TAIL TYPOLOGY ===")

display(
    tail_summary_view[
        [
            "primary_tail_type",
            "n_products",
            "Portfolio Footprint",
            "Total Profit",
            "Net Profit Share",
            "Mean Order Profit",
            "Median CV",
            "Median Loss Rate",
        ]
    ]
    .rename(columns={
        "primary_tail_type": "Tail Type",
        "n_products": "Products",
    })
)

print("\n=== RAW TAIL-MECHANISM OVERLAP AUDIT ===")

display(overlap_audit)
```

  <div style="background:#f1faf5;border-left:3px solid #4f9c70;padding:14px 17px;">
    <div style="font-size:13px;font-weight:700;color:#31714d;margin-bottom:5px;">Tail reading</div>
    <div style="font-size:14px;line-height:1.75;color:#34483c;">
      The 72-product tail separates into three materially different regimes:
      <b>27 Extreme Impact</b> products generate approximately <b>101,539</b> of
      profit and create concentration exposure; <b>19 Extreme Volatility</b>
      products contribute little net profit but carry a strong controllability
      signal; and <b>26 Loss Tail</b> products generate approximately
      <b>−40,529</b> of downside. Raw mechanisms directly explain 70 products;
      the other two are positive Mean Order Profit boundary exceptions whose
      directional assignment is explicitly recorded. No product is silently
      absorbed by a generic fallback.
    </div>

<div style="border-top:1px solid #d9e0e8;padding-top:24px;margin-bottom:24px;"><div style="font-size:12px;font-weight:700;letter-spacing:1px;text-transform:uppercase;color:#7a899b;margin-bottom:6px;">Section 07</div><div style="font-size:29px;font-weight:700;color:#17263c;line-height:1.2;margin-bottom:7px;">Governance Engine</div><div style="font-size:15px;color:#607086;line-height:1.6;">Stress-test the baseline management route and resolve the evidence into one auditable Escalation Status.</div></div>
<p style="font-size:15px;line-height:1.75;color:#2b3440;margin:0 0 14px 0;">Chapter 6 completed the structural classification of the portfolio. Each product now has one <code>analytical_role</code>, describing the recurring economic regime or structural exception that best represents its observed behaviour. Chapter 7 does not replace that classification; it tests how management should act on it.</p>
<p style="font-size:15px;line-height:1.75;color:#2b3440;margin:0 0 14px 0;">The process has four distinct layers. The <b>Analytical Role</b> determines a default <b>Primary Action</b>. Four independent <b>Governance Signals</b> then test whether the evidence still supports that route: <b>Deterioration, Concentration, Tail Overlap, and Classification Instability</b>. A fixed precedence rule converts any overlapping signals into one <b>Escalation Status</b>.</p>
<p style="font-size:15px;line-height:1.75;color:#2b3440;margin:0 0 18px 0;">All rules operate deterministically on the same one-row-per-product table. No product is manually overridden, and the signal evidence remains available after the Escalation Status is resolved.</p>

<div style="font-size:19px;font-weight:700;color:#244e7a;margin:27px 0 9px 0;">7.2 Reference Layer — Threshold Anchors and Classification Instability</div>
<p style="font-size:15px;line-height:1.75;color:#2b3440;margin:0 0 14px 0;">Governance limits are anchored to the <b>1,822-product clustered body</b>, the recurring part of the portfolio. Structural extremes are judged against that reference population and are not allowed to move the benchmarks used to assess them.</p>
<p style="font-size:15px;line-height:1.75;color:#2b3440;margin:0 0 14px 0;">The reference layer supports two different jobs. </tbody></table>
</div>
<div style="font-size:15px;font-weight:700;color:#244e7a;margin:22px 0 8px 0;">Classification Instability: one Governance Signal, two measurement systems</div>
<table style="width:100%;border-collapse:collapse;font-size:13.2px;color:#263444;margin-bottom:16px;"><thead><tr style="background:#17263c;color:white;"><th style="padding:9px 11px;text-align:left;border:1px solid #17263c;width:20%;">Portfolio Domain</th><th style="padding:9px 11px;text-align:left;border:1px solid #17263c;width:37%;">How Confidence Is Measured</th><th style="padding:9px 11px;text-align:left;border:1px solid #17263c;">Interpretation and Stored Output</th></tr></thead><tbody>
<tr><td style="padding:10px 11px;border:1px solid #dce4ee;"><b>Clustered Body</b></td><td style="padding:10px 11px;border:1px solid #dce4ee;">Relative distance margin between the assigned centroid and the nearest alternative centroid.</td><td style="padding:10px 11px;border:1px solid #dce4ee;">A small margin means the assignment is borderline. Stored in <code>assignment_instability</code> as <b>none</b>, <b>watch</b>, or <b>high</b>.</td></tr>
<tr style="background:#f8fafc;"><td style="padding:10px 11px;border:1px solid #dce4ee;"><b>Structural Tail</b></td><td style="padding:10px 11px;border:1px solid #dce4ee;">Distance beyond the profit, volatility, or impact reference that created the primary tail role.</td><td style="padding:10px 11px;border:1px solid #dce4ee;">A small excess means the product only narrowly entered that tail regime. The same <code>assignment_instability</code> levels are used.</td></tr>
</tbody></table>
<div style="background:#fffaf0;border-left:3px solid #b58a37;padding:13px 16px;margin:18px 0;"><div style="font-size:13px;font-weight:700;color:#725821;margin-bottom:5px;">What Classification Instability does—and does not—mean</div><div style="font-size:14px;line-height:1.7;color:#4a4438;">This signal measures <b>confidence in the assigned role</b>, not whether the product is profitable or attractive. Weak anchoring can justify review even when current economics are strong; poor economics can exist without instability when the product is securely located in a loss-oriented role.</div></div>
</div>


```python
# =============================================================================
# 7.2 BODY-ANCHORED REFERENCES AND CLASSIFICATION INSTABILITY SIGNAL
# =============================================================================

body_ref = chapter_07.body_ref
threshold_export = chapter_07.threshold_export
body_instability = chapter_07.body_instability

print("\n=== GOVERNANCE THRESHOLD REFERENCE ===")

display(threshold_export.round({"value": 4}))

print("\n=== CLASSIFICATION INSTABILITY GOVERNANCE SIGNAL ===")

display(
    chapter_07.flow_df["assignment_instability"]
    .value_counts()
    .rename_axis("Classification Instability")
    .reset_index(name="Products")
)
```

<div style="font-family:Arial,sans-serif;max-width:1150px;margin:0 auto;">
<div style="font-size:19px;font-weight:700;color:#244e7a;margin:27px 0 9px 0;">7.3 Four Governance Signals — Exact Evaluation Logic</div>
<p style="font-size:15px;line-height:1.75;color:#2b3440;margin:0 0 14px 0;">The Governance Signals are independent diagnostic tests. A product can activate more than one because each signal asks a different question: whether expected economics have weakened, whether too much value depends on the product, whether several tail mechanisms coexist, and whether the assigned role is secure.</p>
<div style="background:#f7f9fc;border:1px solid #dce4ee;border-radius:8px;padding:15px 17px;margin:18px 0;">
<div style="font-size:12px;font-weight:700;letter-spacing:.8px;text-transform:uppercase;color:#66788e;margin-bottom:10px;">Complete Governance-Signal Framework</div>
<table style="width:100%;border-collapse:collapse;font-size:13px;color:#263444;"><thead><tr style="background:#17263c;color:white;"><th style="padding:9px 10px;text-align:left;border:1px solid #17263c;width:18%;">Governance Signal</th><th style="padding:9px 10px;text-align:left;border:1px solid #17263c;width:29%;">What It Tests</th><th style="padding:9px 10px;text-align:left;border:1px solid #17263c;width:27%;">Evaluation Basis</th><th style="padding:9px 10px;text-align:left;border:1px solid #17263c;width:26%;">Stored Output</th></tr></thead><tbody>
<tr><td style="padding:10px;border:1px solid #dce4ee;"><b>Deterioration</b></td><td style="padding:10px;border:1px solid #dce4ee;">Whether profit, volatility, or loss frequency breach the tolerance of the product's own role.</td><td style="padding:10px;border:1px solid #dce4ee;">Seven role-specific moderate/severe rule sets.</td><td style="padding:10px;border:1px solid #dce4ee;"><code>deterioration_<wbr>level</code></td></tr>
<tr style="background:#f8fafc;"><td style="padding:10px;border:1px solid #dce4ee;"><b>Concentration</b></td><td style="padding:10px;border:1px solid #dce4ee;">Whether disproportionate positive contribution creates portfolio dependency.</td><td style="padding:10px;border:1px solid #dce4ee;">Impact roles only; body Profit Scale p95/p99 and deterioration interaction.</td><td style="padding:10px;border:1px solid #dce4ee;"><code>concentration_<wbr>level</code></td></tr>
<tr><td style="padding:10px;border:1px solid #dce4ee;"><b>Tail Overlap</b></td><td style="padding:10px;border:1px solid #dce4ee;">Whether one structural-tail product carries multiple exception mechanisms.</td><td style="padding:10px;border:1px solid #dce4ee;"><code>n_tail_flags ≥ 2</code>.</td><td style="padding:10px;border:1px solid #dce4ee;"><code>compound_<wbr>overlap_<wbr>flag</code></td></tr>
<tr style="background:#f8fafc;"><td style="padding:10px;border:1px solid #dce4ee;"><b>Classification Instability</b></td><td style="padding:10px;border:1px solid #dce4ee;">Whether the product is weakly anchored or sits close to a classification boundary.</td><td style="padding:10px;border:1px solid #dce4ee;">Body centroid margin or tail boundary distance.</td><td style="padding:10px;border:1px solid #dce4ee;"><code>assignment_<wbr>instability</code></td></tr>
</tbody></table>
</div>
<div style="font-size:15px;font-weight:700;color:#244e7a;margin:22px 0 8px 0;">Signal 1 — Deterioration: role-relative tolerances</div>
<p style="font-size:15px;line-height:1.75;color:#2b3440;margin:0 0 14px 0;">The same KPI value can mean different things in different roles. A CV of 1.5 is unusual for Stable Profit but ordinary inside a volatility-oriented regime. Each product is therefore tested only against the rule set attached to its own <code>analytical_role</code>.</p>
<div style="background:#f7f9fc;border:1px solid #dce4ee;border-radius:8px;padding:15px 17px;margin:18px 0;overflow-x:auto;">
<table style="width:100%;border-collapse:collapse;font-size:12.7px;color:#263444;min-width:1000px;"><thead><tr style="background:#17263c;color:white;"><th style="padding:9px;text-align:left;border:1px solid #17263c;">Analytical Role</th><th style="padding:9px;text-align:left;border:1px solid #17263c;">Moderate Deterioration — any condition</th><th style="padding:9px;text-align:left;border:1px solid #17263c;">Severe Deterioration — any condition</th></tr></thead><tbody>
<tr><td style="padding:9px;border:1px solid #dce4ee;"><b>Stable Profit</b></td><td style="padding:9px;border:1px solid #dce4ee;">CV &gt; 1.00 or Loss Rate &gt; 30%</td><td style="padding:9px;border:1px solid #dce4ee;">Mean Profit ≤ 0 or CV &gt; 1.50 or Loss Rate &gt; 45%</td></tr>
<tr style="background:#f8fafc;"><td style="padding:9px;border:1px solid #dce4ee;"><b>High-Impact Profit</b></td><td style="padding:9px;border:1px solid #dce4ee;">CV &gt; 1.25 or Loss Rate &gt; 25%</td><td style="padding:9px;border:1px solid #dce4ee;">Mean Profit ≤ 0 or CV &gt; 2.00 or Loss Rate &gt; 40%</td></tr>
<tr><td style="padding:9px;border:1px solid #dce4ee;"><b>Risk-Oriented Profit</b></td><td style="padding:9px;border:1px solid #dce4ee;">CV &gt; 1.50 or Loss Rate &gt; 35%</td><td style="padding:9px;border:1px solid #dce4ee;">Mean Profit ≤ 0 or CV &gt; 3.00 or Loss Rate &gt; 50%</td></tr>
<tr style="background:#f8fafc;"><td style="padding:9px;border:1px solid #dce4ee;"><b>Stable Loss</b></td><td style="padding:9px;border:1px solid #dce4ee;">CV &gt; 1.50 or Loss Rate &gt; 45%</td><td style="padding:9px;border:1px solid #dce4ee;">CV &gt; 2.50 or Loss Rate &gt; 60%</td></tr>
<tr><td style="padding:9px;border:1px solid #dce4ee;"><b>Extreme Impact Tail</b></td><td style="padding:9px;border:1px solid #dce4ee;">CV &gt; 1.50 or Loss Rate &gt; 25%</td><td style="padding:9px;border:1px solid #dce4ee;">Mean Profit ≤ 0 or CV &gt; 2.50 or Loss Rate &gt; 40%</td></tr>
<tr style="background:#f8fafc;"><td style="padding:9px;border:1px solid #dce4ee;"><b>Extreme Volatility Tail</b></td><td style="padding:9px;border:1px solid #dce4ee;">CV &gt; body p95 or Loss Rate &gt; body p90</td><td style="padding:9px;border:1px solid #dce4ee;">CV &gt; body p99 or Mean Profit ≤ 0</td></tr>
<tr><td style="padding:9px;border:1px solid #dce4ee;"><b>Loss Tail</b></td><td style="padding:9px;border:1px solid #dce4ee;">CV &gt; body p90 or Loss Rate &gt; 50%</td><td style="padding:9px;border:1px solid #dce4ee;">CV &gt; body p95 or Loss Rate &gt; 65%</td></tr>
</tbody></table>
</div>
<div style="background:#fffaf0;border-left:3px solid #b58a37;padding:13px 16px;margin:18px 0;"><div style="font-size:13px;font-weight:700;color:#725821;margin-bottom:5px;">How the deterioration matrix is executed</div><div style="font-size:14px;line-height:1.7;color:#4a4438;">Conditions inside each severity level use <b>OR logic</b>: one breach is sufficient. Severe evidence overrides moderate evidence. Mean Profit ≤ 0 is not treated as a new severe event for Stable Loss or Loss Tail because negative economics already define those roles; severity there is driven by volatility and loss frequency.</div></div>
<div style="font-size:15px;font-weight:700;color:#244e7a;margin:22px 0 8px 0;">Signals 2–4 — compact exact rules</div>
<table style="width:100%;border-collapse:collapse;font-size:13.2px;color:#263444;margin-bottom:16px;"><thead><tr style="background:#17263c;color:white;"><th style="padding:9px 11px;text-align:left;border:1px solid #17263c;width:22%;">Governance Signal</th><th style="padding:9px 11px;text-align:left;border:1px solid #17263c;width:45%;">Exact Trigger</th><th style="padding:9px 11px;text-align:left;border:1px solid #17263c;">Why It Matters</th></tr></thead><tbody>
<tr><td style="padding:10px 11px;border:1px solid #dce4ee;"><b>Concentration</b></td><td style="padding:10px 11px;border:1px solid #dce4ee;">Only High-Impact Profit and Extreme Impact Tail are eligible. <b>Elevated</b>: Profit Scale ≥ body p95. <b>Extreme</b>: Profit Scale ≥ body p99, or ≥ p95 together with any Deterioration.</td><td style="padding:10px 11px;border:1px solid #dce4ee;">Separates valuable contribution from portfolio dependency that needs stronger control.</td></tr>
<tr style="background:#f8fafc;"><td style="padding:10px 11px;border:1px solid #dce4ee;"><b>Tail Overlap</b></td><td style="padding:10px 11px;border:1px solid #dce4ee;">Active when a tail product carries at least two raw mechanisms: <code>n_tail_flags ≥ 2</code>.</td><td style="padding:10px 11px;border:1px solid #dce4ee;">Multiple exception mechanisms weaken a single clean tail interpretation.</td></tr>
<tr><td style="padding:10px 11px;border:1px solid #dce4ee;"><b>Classification Instability</b></td><td style="padding:10px 11px;border:1px solid #dce4ee;"><b>Watch</b> or <b>High</b> according to the body centroid margin or tail boundary-distance rules established in 7.2.</td><td style="padding:10px 11px;border:1px solid #dce4ee;">Distinguishes a secure assignment from a borderline classification that may need review.</td></tr>
</tbody></table>
<div style="background:#f1faf5;border-left:3px solid #4f9c70;padding:12px 14px;margin:16px 0;"><div style="font-size:13px;font-weight:700;color:#31714d;margin-bottom:5px;">Independent by design</div><div style="font-size:14px;line-height:1.7;color:#34483c;">The four Governance Signals are not forced into mutually exclusive classes. <code>n_active_signals</code> records how many signal families are active, while <code>primary_signal</code> preserves the highest-priority reason used in the resolution layer.</div></div>
</div>


```python
# =============================================================================
# 7.3 EXECUTE THE DETERIORATION AND CONCENTRATION GOVERNANCE SIGNALS
# =============================================================================

deterioration_specs = chapter_07.deterioration_specs

print("\n=== EXECUTED DETERIORATION RULE SPECIFICATION ===")

display(deterioration_specs.round(4))
```

<div style="font-family:Arial,sans-serif;max-width:1150px;margin:0 auto;">
<div style="font-size:19px;font-weight:700;color:#244e7a;margin:27px 0 9px 0;">7.4 Governance-Signal Precedence and Escalation Status</div>
<p style="font-size:15px;line-height:1.75;color:#2b3440;margin:0 0 14px 0;">The four Governance Signals are evaluated independently, so several can be active for the same product. Management nevertheless needs one resolved control consequence. Ordered precedence therefore converts the combined evidence into exactly one <code>escalation_status</code>.</p>
<div style="background:#f7f9fc;border:1px solid #dce4ee;border-radius:8px;padding:15px 17px;margin:18px 0;">
<table style="width:100%;border-collapse:collapse;font-size:13.3px;color:#263444;"><thead><tr style="background:#17263c;color:white;"><th style="padding:9px 11px;text-align:center;border:1px solid #17263c;width:9%;">Priority</th><th style="padding:9px 11px;text-align:left;border:1px solid #17263c;width:24%;">Escalation Status</th><th style="padding:9px 11px;text-align:left;border:1px solid #17263c;width:35%;">Governance-Signal Trigger</th><th style="padding:9px 11px;text-align:left;border:1px solid #17263c;">Resolved Meaning</th></tr></thead><tbody>
<tr><td style="padding:10px;border:1px solid #dce4ee;text-align:center;"><b>1</b></td><td style="padding:10px;border:1px solid #dce4ee;"><b>Reassess Classification</b></td><td style="padding:10px;border:1px solid #dce4ee;">Tail Overlap or <b>High</b> Classification Instability.</td><td style="padding:10px;border:1px solid #dce4ee;">The classification itself may be unreliable; review it before routing action.</td></tr>
<tr style="background:#f8fafc;"><td style="padding:10px;border:1px solid #dce4ee;text-align:center;"><b>2</b></td><td style="padding:10px;border:1px solid #dce4ee;"><b>Reroute Action</b></td><td style="padding:10px;border:1px solid #dce4ee;">Severe Deterioration or Extreme Concentration.</td><td style="padding:10px;border:1px solid #dce4ee;">The role remains trusted, but the baseline Primary Action no longer fits observed behaviour.</td></tr>
<tr><td style="padding:10px;border:1px solid #dce4ee;text-align:center;"><b>3</b></td><td style="padding:10px;border:1px solid #dce4ee;"><b>Intensify Action</b></td><td style="padding:10px;border:1px solid #dce4ee;">Moderate Deterioration, Elevated Concentration, or <b>Watch</b> Classification Instability.</td><td style="padding:10px;border:1px solid #dce4ee;">The baseline route still holds, but it requires tighter control.</td></tr>
<tr style="background:#f8fafc;"><td style="padding:10px;border:1px solid #dce4ee;text-align:center;"><b>4</b></td><td style="padding:10px;border:1px solid #dce4ee;"><b>No Escalation</b></td><td style="padding:10px;border:1px solid #dce4ee;">No active Governance Signal.</td><td style="padding:10px;border:1px solid #dce4ee;">The Primary Action holds without additional escalation.</td></tr>
</tbody></table>
</div>
<div style="background:#fffaf0;border-left:3px solid #b58a37;padding:13px 16px;margin:18px 0;"><div style="font-size:13px;font-weight:700;color:#725821;margin-bottom:5px;">Two outputs are retained</div><div style="font-size:14px;line-height:1.7;color:#4a4438;"><code>primary_signal</code> records the dominant governance reason for audit and explanation. <code>escalation_status</code> records the resolved management consequence. A higher-priority status overwrites a lower one, but the underlying signal fields remain unchanged.</div></div>
</div>


```python
# =============================================================================
# 7.4 RESOLVE FOUR GOVERNANCE SIGNALS INTO ONE ESCALATION STATUS
# =============================================================================

signal_summary = chapter_07.signal_summary
escalation_summary = chapter_07.escalation_summary

print("\n=== GOVERNANCE SIGNAL SUMMARY ===")

display(signal_summary.round(3))

print("\n=== RESOLVED ESCALATION STATUS ===")

display(escalation_summary.round(3))
```

    No escalation preserves the baseline. Moderate pressure tightens it. Severe
    failure redirects it. Classification ambiguity pauses ordinary routing and
    sends the case to reassessment.
  </p>
</div>

<div style="font-family:Arial,sans-serif;max-width:1150px;margin:0 auto;">
  <div style="font-size:19px;font-weight:700;color:#244e7a;margin:27px 0 9px 0;">9.1 Role-Relative Performance</div>
  <p style="font-size:15px;line-height:1.75;color:#2b3440;margin:0 0 12px 0;">A product is not labelled strong or weak against one portfolio-wide benchmark. It is compared with products in the <b>same analytical role</b>, because the normal profitability level of a Stable Profit product is fundamentally different from that of a Loss Tail or Extreme Impact product.</p>
  <p style="font-size:15px;line-height:1.75;color:#2b3440;margin:0 0 14px 0;">The comparison uses each role's distribution of <b>Mean Order Profit</b>. The percentile checks are evaluated in the order shown below, so p90 is tested before p75.</p>
  <div style="background:#f7f9fc;border:1px solid #dce4ee;border-radius:8px;padding:15px 17px;margin:14px 0 16px 0;">
    <table style="width:100%;border-collapse:collapse;font-size:13.3px;color:#263444;">

      <tbody>
        <tr><td style="padding:9px 11px;border:1px solid #dce4ee;"><b>High Positive Deviation</b></td><td style="padding:9px 11px;border:1px solid #dce4ee;"><code>mean_order_profit ≥ role p90</code></td><td style="padding:9px 11px;border:1px solid #dce4ee;">Top 10% performer inside its own role.</td></tr>
        <tr style="background:#f8fafc;"><td style="padding:9px 11px;border:1px solid #dce4ee;"><b>Positive Deviation</b></td><td style="padding:9px 11px;border:1px solid #dce4ee;"><code>role p75 ≤ mean_order_profit &lt; role p90</code></td><td style="padding:9px 11px;border:1px solid #dce4ee;">Upper-quartile performer, but not top decile.</td></tr>
        <tr><td style="padding:9px 11px;border:1px solid #dce4ee;"><b>Negative Deviation</b></td><td style="padding:9px 11px;border:1px solid #dce4ee;"><code>mean_order_profit ≤ role p25</code></td><td style="padding:9px 11px;border:1px solid #dce4ee;">Bottom-quartile performer inside its own role.</td></tr>
        <tr style="background:#f8fafc;"><td style="padding:9px 11px;border:1px solid #dce4ee;"><b>Role-Aligned</b></td><td style="padding:9px 11px;border:1px solid #dce4ee;"><code>role p25 &lt; mean_order_profit &lt; role p75</code></td><td style="padding:9px 11px;border:1px solid #dce4ee;">Performance remains inside the role's central operating range.</td></tr>

      <div style="font-size:13.5px;line-height:1.7;color:#34483c;"><code>flow_df[primary_action, escalation_status]</code> → <code>final_intervention_map</code> → <code>flow_df[final_intervention]</code> → <code>governance_scorecard</code> → <code>governance_scorecard_sensitivity</code>. Chapter 9 enriches this same canonical <code>flow_df</code> with a separate economic-priority lens; it does not overwrite the resolved intervention route.</div>

    <div style="font-size:14.5px;line-height:1.75;color:#334155;">This framework is a <b>static, in-sample governance diagnostic</b>. It identifies where management should look first by combining portfolio structure, governance status, economic exposure, and robustness evidence. It does not predict future outcomes, establish causality, or replace commercial judgment.</div>

<div style="font-family:Arial,sans-serif;max-width:1150px;margin:0 auto;">
  <div style="font-size:19px;font-weight:700;color:#244e7a;margin:27px 0 9px 0;">10.1 Supported Use and Claim Limits</div>
  <p style="font-size:14.5px;line-height:1.75;color:#2b3440;margin:0 0 14px 0;">The framework is designed as a structured management-review and governance-triage layer. It supports the following uses:</p>

    <table style="width:100%;border-collapse:collapse;font-size:13.3px;color:#263444;">

        </tr>
      </thead>

    <div style="font-size:13px;font-weight:700;color:#725821;margin-bottom:6px;">Claim Limits</div>
    <div style="font-size:13.7px;line-height:1.75;color:#4a4438;">The framework does not establish uniquely correct clusters, future persistence, causal effects, universally optimal thresholds, automatic intervention outcomes, or complete commercial feasibility. It also does not justify ignoring pricing, demand, inventory, suppliers, seasonality, operating costs, or strategic context.</div>
  </div>
  <div style="background:#eef6fc;border-left:3px solid #5b8db8;padding:11px 14px;font-size:13.5px;line-height:1.7;color:#334155;"><b>Correct reading:</b> the framework is strongest as a <b>governance triage system</b>. It narrows attention to the products and regimes where value, risk, and control misalignment are most concentrated. It does not decide automatically; it shows where managerial review should concentrate first.</div>

<div style="font-family:Arial,sans-serif;max-width:1150px;margin:0 auto;">
  <div style="font-size:19px;font-weight:700;color:#244e7a;margin:27px 0 9px 0;">10.2 Methodological Boundary</div>
  <p style="font-size:14.5px;line-height:1.75;color:#2b3440;margin:0 0 14px 0;">Roles, thresholds, escalation states, and opportunity classes are derived from the same historical portfolio snapshot. This creates an unavoidable <b>in-sample calibration boundary</b>. That boundary is appropriate here because the objective is to build a deterministic, auditable control surface for the observed portfolio—not a forecasting model.</p>
  <div style="background:#f7f9fc;border:1px solid #dce4ee;border-radius:8px;padding:14px 16px;margin:0 0 15px 0;">
    <div style="font-size:13px;font-weight:700;letter-spacing:0.6px;text-transform:uppercase;color:#66788e;margin-bottom:8px;">Robustness Checks</div>
    <table style="width:100%;border-collapse:collapse;font-size:13.3px;color:#263444;">
      <thead><tr style="background:#eaf2f7;color:#244e7a;"><th style="padding:9px 11px;text-align:left;border:1px solid #cfdde8;width:32%;">Validation Layer</th><th style="padding:9px 11px;text-align:left;border:1px solid #cfdde8;">Purpose</th></tr></thead>
      <tbody>
        <tr><td style="padding:9px 11px;border:1px solid #dce4ee;"><b>Body/tail boundary sensitivity</b></td><td style="padding:9px 11px;border:1px solid #dce4ee;">Tests whether the clusterable body depends on one arbitrary empirical cutoff.</td></tr>
        <tr style="background:#f8fafc;"><td style="padding:9px 11px;border:1px solid #dce4ee;"><b>CV-support sensitivity</b></td><td style="padding:9px 11px;border:1px solid #dce4ee;">Tests whether non-estimable volatility handling drives the K-means body taxonomy.</td></tr>
        <tr><td style="padding:9px 11px;border:1px solid #dce4ee;"><b>Cluster resampling stability</b></td><td style="padding:9px 11px;border:1px solid #dce4ee;">Tests whether the body segmentation remains stable under sampling variation.</td></tr>
        <tr style="background:#f8fafc;"><td style="padding:9px 11px;border:1px solid #dce4ee;"><b>Governance scorecard sensitivity</b></td><td style="padding:9px 11px;border:1px solid #dce4ee;">Tests whether hard exposure remains material under stricter or looser escalation thresholds.</td></tr>
        <tr><td style="padding:9px 11px;border:1px solid #dce4ee;"><b>Hidden-opportunity threshold stress test</b></td><td style="padding:9px 11px;border:1px solid #dce4ee;">Tests whether the opportunity signal remains economically material under alternative reliability filters.</td></tr>
      </tbody>
    </table>
  </div>
  <div style="background:#f1faf5;border-left:3px solid #4f9c70;padding:12px 15px;color:#34483c;font-size:13.7px;line-height:1.75;"><b>Boundary reading:</b> these checks reduce specification risk, but they do not turn the framework into a predictive or causal model. They support the narrower claim that <b>the analytical diagnosis is stable enough to support structured management review under the observed portfolio snapshot</b>.</div>

<div style="font-family:Arial,sans-serif;max-width:1150px;margin:0 auto;">

  <div style="font-size:19px;font-weight:700;color:#244e7a;margin:27px 0 9px 0;">10.3 Productionisation and Deployment Path</div>

  <p style="font-size:14.5px;line-height:1.75;color:#2b3440;margin:0 0 12px 0;">The current framework operates on one historical portfolio snapshot. A production extension would convert the same analytical logic into a <b>scheduled, version-controlled management cycle</b>: ingest new transactions, rebuild the product evidence, apply the governed decision rules, publish the resulting interventions, and monitor what changes after management acts.</p>

  <p style="font-size:14.5px;line-height:1.75;color:#2b3440;margin:0 0 17px 0;">The production objective would not be to automate commercial decisions. It would be to ensure that every reporting cycle produces a consistent and auditable answer to three questions: <b>what changed, which management routes remain valid, and where attention should move next</b>.</p>

  <div style="background:#f7f9fc;border:1px solid #dce4ee;border-radius:8px;padding:15px 16px;margin:0 0 18px 0;">

  <div style="background:#eef6fc;border:1px solid #cfe0ed;border-left:4px solid #2563eb;border-radius:8px;padding:14px 17px;margin:0 0 18px 0;">
    <div style="font-size:13px;font-weight:700;color:#1d4ed8;margin-bottom:6px;">Critical Model-Control Rule — Score Consistently, Recalibrate Deliberately</div>
    <div style="font-size:13.8px;line-height:1.75;color:#334155;">K-means should not be refitted automatically during every production run. Routine scoring should reuse the approved scaler, cluster centroids, cluster-to-role mapping, body/tail references, and governance rules. Otherwise, cluster definitions could move between reporting periods and create apparent role changes that reflect model movement rather than genuine economic change.</div>
    <div style="font-size:13.8px;line-height:1.75;color:#334155;margin-top:8px;">Recalibration should take place through a separate controlled review—either at an agreed interval or when monitoring identifies material portfolio drift. A candidate specification should first run as a <b>challenger</b>, with its stability, interpretation, economic coverage, and decision impact compared with the active production version before approval.</div>
  </div>

  <div style="font-size:16px;font-weight:700;color:#244e7a;margin:22px 0 9px 0;">Required Production Controls</div>

  <div style="background:#f7f9fc;border:1px solid #dce4ee;border-radius:8px;padding:14px 16px;margin:0 0 17px 0;">

        </tr>
        <tr style="background:#f8fafc;">

        </tr>
        <tr>

          <td style="padding:9px 11px;border:1px solid #dce4ee;">Retain explicit support flags for new or sparsely observed products and route them to monitored review when the evidence is insufficient for a fully trusted classification.</td>

        <tr>
          <td style="padding:9px 11px;border:1px solid #dce4ee;"><b>Commercial validation gate</b></td>
          <td style="padding:9px 11px;border:1px solid #dce4ee;">Review pricing, demand, inventory, suppliers, seasonality, operating costs, and strategic commitments before approving scaling, restructuring, or exit decisions.</td>
        </tr>
        <tr style="background:#f8fafc;">
          <td style="padding:9px 11px;border:1px solid #dce4ee;"><b>Outcome monitoring</b></td>
          <td style="padding:9px 11px;border:1px solid #dce4ee;">Compare approved interventions with subsequent profit, volatility, loss frequency, role movement, and exposure to determine whether the action produced the intended result.</td>
        </tr>
      </tbody>
    </table>

  <div style="background:#fffaf0;border:1px solid #eadcb9;border-left:4px solid #d99a2b;border-radius:8px;padding:14px 17px;margin:0 0 18px 0;">
    <div style="font-size:13px;font-weight:700;color:#725821;margin-bottom:6px;">How Repeated Snapshots Change the Interpretation</div>
    <div style="font-size:13.8px;line-height:1.75;color:#4a4438;">Within the current single snapshot, <b>Deterioration</b> means that a product breaches the economic tolerances of its assigned role. It does not establish that the product declined through time. Once the same governed snapshot is produced repeatedly, the system can begin to distinguish a temporary breach from persistent deterioration, genuine role migration, sustained recovery, or repeated management-route failure.</div>
  </div>

  <div style="font-size:16px;font-weight:700;color:#244e7a;margin:22px 0 9px 0;">Controlled Rollout</div>

  <div style="display:grid;grid-template-columns:repeat(3,1fr);gap:12px;margin:0 0 18px 0;">
    <div style="background:#f7f9fc;border:1px solid #dce4ee;border-top:4px solid #5b8db8;border-radius:8px;padding:14px 15px;">
      <div style="font-size:12px;font-weight:700;letter-spacing:0.6px;text-transform:uppercase;color:#46769e;margin-bottom:5px;">Phase 1 · Shadow Mode</div>
      <div style="font-size:13.4px;line-height:1.7;color:#334155;">Run several complete reporting cycles without changing operational decisions. Validate data reliability, assignment stability, refresh behaviour, and the usefulness of the generated alerts.</div>
    </div>
    <div style="background:#f7f9fc;border:1px solid #dce4ee;border-top:4px solid #d99a2b;border-radius:8px;padding:14px 15px;">

      <div style="font-size:13.4px;line-height:1.7;color:#334155;">Use the Priority Zones with a defined commercial or product team. Record accepted, rejected, and modified recommendations together with the business reason.</div>
    </div>
    <div style="background:#f7f9fc;border:1px solid #dce4ee;border-top:4px solid #4f9c70;border-radius:8px;padding:14px 15px;">
      <div style="font-size:12px;font-weight:700;letter-spacing:0.6px;text-transform:uppercase;color:#31714d;margin-bottom:5px;">Phase 3 · Production Control</div>
      <div style="font-size:13.4px;line-height:1.7;color:#334155;">Activate scheduled runs, governed Power BI refreshes, exception review, decision logging, outcome monitoring, and controlled model or threshold recalibration.</div>
    </div>
  </div>

  <div style="background:#f1faf5;border-left:3px solid #4f9c70;padding:14px 17px;margin-bottom:18px;">
    <div style="font-size:13px;font-weight:700;color:#31714d;margin-bottom:5px;">Final Interpretation</div>
    <div style="font-size:14px;line-height:1.75;color:#34483c;">The production target is a <b>human-governed portfolio control cycle</b>. The analytical engine identifies the relevant route, active signals, and priority context; management validates the commercial evidence and owns the final action. This preserves consistency and auditability without turning the framework into an autonomous decision engine.</div>
  </div>

    <div style="font-size:13.7px;line-height:1.75;color:#334155;">The framework remains appropriately bounded as <b>static</b>, <b>in-sample</b>, <b>diagnostic</b>, <b>auditable</b>, and <b>non-autonomous</b> in its current form. Its production path is nevertheless explicit: repeated governed snapshots, stable reference scoring, versioned rules, management approval, outcome tracking, and controlled recalibration can extend the analysis into a repeatable portfolio decision system.</div>
